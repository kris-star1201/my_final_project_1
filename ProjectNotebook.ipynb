# PA School Progress Tracker

This project helps students interested in applying to Physician Assistant (PA) school compare
their current qualifications to the admission requirements of one or more PA programs.

This program helps students understand how their current GPA, science GPA, and patient care experience (PCE) hours compare to the requirements of PA schools. Whether looking at just one school or several, it shows what they’re on track with and what areas might need some extra attention, making the journey to their goals a bit clearer and more manageable.

Example 2 demonstrates comparing a student's qualifications to multiple PA programs using a list of 
dictionaries and a loop.
The output displays the remaining GPA and PCE requirements for each school, allowing students to compare their progress across multiple PA programs at once. 

```python
school = [
    {
        "School": "Example PA Program",
        "Min_GPA": 3.2,
        "Min_Science_GPA": 3.0,
        "Required_PCE_hours": 1000
    },
    { 
        "School": "More Competitive PA Program",
        "Min_GPA": 3.5,
        "Min_Science_GPA": 3.3,
        "Required_PCE_hours": 2000
    },
    { 
        "School": "Lower PCE Requirement Program",
        "Min_GPA": 3.0,
        "Min_Science_GPA": 3.0,
        "Required_PCE_hours": 500
    }
]

results = module1.compare_schools(
    3.0,
    2.8,
    500,
    school
)

for result in results:
    print(result)
```
